```python
import numpy as np



# A force of 100 N at 30° above the horizontal — resolve into components

F = 100.0

theta = np.radians(30)

Fx = F * np.cos(theta)

Fy = F * np.sin(theta)

print(f"Fx = {Fx:.1f} N,  Fy = {Fy:.1f} N")
```

```python
# Resultant of two forces

F1 = np.array([Fx, Fy])

F2 = np.array([-40.0, 10.0])

R = F1 + F2

print(f"Resultant R = {R.round(1)} N,  |R| = {np.linalg.norm(R):.1f} N")
```

```python
import matplotlib.pyplot as plt



plt.quiver(0, 0, F1[0], F1[1], angles="xy", scale_units="xy", scale=1, color="tab:blue", label="F1")

plt.quiver(0, 0, F2[0], F2[1], angles="xy", scale_units="xy", scale=1, color="tab:red", label="F2")

plt.quiver(0, 0, R[0], R[1], angles="xy", scale_units="xy", scale=1, color="tab:green", label="R")

plt.xlim(-60, 110)

plt.ylim(-10, 80)

plt.axhline(0, color="black", linewidth=0.5)

plt.axvline(0, color="black", linewidth=0.5)

plt.grid(True)

plt.legend()

plt.title("Resultant of two forces")

plt.show()
```
