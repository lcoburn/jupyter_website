```python
# SI prefixes: expressing a wavelength in metres
wavelength_nm = 650          # red light, in nanometres
wavelength_m = wavelength_nm * 1e-9
print(f"{wavelength_nm} nm = {wavelength_m} m = {wavelength_m:.2e} m")
```

```python
# Checking units in an equation: E = m*c^2 gives joules (kg·m²/s²)
m = 1.0        # kg
c = 3.00e8     # m/s
E = m * c**2
print(f"Energy of 1 kg = {E:.2e} J")
```

```python
import numpy as np
import matplotlib.pyplot as plt

prefixes = {"nm": 1e-9, "µm": 1e-6, "mm": 1e-3, "m": 1.0, "km": 1e3}
values_m = [wavelength_m / f for f in prefixes.values()]

plt.bar(prefixes.keys(), values_m)
plt.yscale("log")
plt.ylabel("650 nm expressed in each unit")
plt.title("Same quantity, different units (log scale)")
plt.show()
```
