# Finding proportional relationships by linearising graphs

A straight-line graph through the origin shows that one plotted quantity is directly proportional to the other.

This notebook shows:
1. $y \propto x$
2. $y \propto x^2$
3. $y \propto \sin x$


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Direct proportion: $y \propto x$

Take $y=3x$. Plotting $y$ against $x$ gives a straight line through the origin.


```python
x1 = np.arange(0, 7)
y1 = 3*x1

display(pd.DataFrame({'x': x1, 'y': y1}))

plt.figure(figsize=(7,5))
plt.plot(x1, y1, 'o-')
plt.xlabel('x')
plt.ylabel('y')
plt.title('y against x')
plt.grid(alpha=0.3)
plt.show()

print('Straight line through the origin → y is proportional to x')
```

## 2. Quadratic relationship: $y \propto x^2$

Take $y=x^2$. The graph of $y$ against $x$ is curved, but the graph of $y$ against $x^2$ is a straight line.


```python
x2 = np.arange(0, 7)
y2 = x2**2
x2_sq = x2**2

display(pd.DataFrame({'x': x2, 'x²': x2_sq, 'y': y2}))

fig, axes = plt.subplots(1, 2, figsize=(12,4.5))

axes[0].plot(x2, y2, 'o-')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].set_title('y against x')
axes[0].grid(alpha=0.3)

axes[1].plot(x2_sq, y2, 'o-')
axes[1].set_xlabel('x²')
axes[1].set_ylabel('y')
axes[1].set_title('y against x²')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print('Straight y against x² → y is proportional to x²')
```

## 3. Sine relationship: $y \propto \sin x$

Take $y=4\sin x$. The graph of $y$ against $x$ is sinusoidal, but the graph of $y$ against $\sin x$ is a straight line.


```python
x_deg = np.arange(0, 361, 30)
x_rad = np.deg2rad(x_deg)
sin_x = np.sin(x_rad)
y3 = 4*sin_x

display(pd.DataFrame({
    'x (degrees)': x_deg,
    'sin(x)': np.round(sin_x, 3),
    'y': np.round(y3, 3)
}))

fig, axes = plt.subplots(1, 2, figsize=(12,4.5))

axes[0].plot(x_deg, y3, 'o-')
axes[0].set_xlabel('x (degrees)')
axes[0].set_ylabel('y')
axes[0].set_title('y against x')
axes[0].grid(alpha=0.3)

axes[1].plot(sin_x, y3, 'o')
axes[1].set_xlabel('sin(x)')
axes[1].set_ylabel('y')
axes[1].set_title('y against sin(x)')
axes[1].grid(alpha=0.3)

s = np.linspace(-1, 1, 200)
axes[1].plot(s, 4*s, '--')

plt.tight_layout()
plt.show()

print('Straight y against sin(x) → y is proportional to sin(x)')
```

## Main idea

We transform one axis to hunt for a straight line:

- straight $y$ against $x$ → $y \propto x$
- straight $y$ against $x^2$ → $y \propto x^2$
- straight $y$ against $\sin x$ → $y \propto \sin x$
