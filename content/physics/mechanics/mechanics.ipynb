```python
# Free fall from rest: how far and how fast after t seconds?

g = 9.81   # m/s^2

t = 3.0    # s



v = g * t

h = 0.5 * g * t**2

print(f"After {t} s: speed = {v:.1f} m/s, distance fallen = {h:.1f} m")
```

```python
# Newton's second law: F = m*a

m = 1200.0   # kg (a car)

a = 2.5      # m/s^2

F = m * a

print(f"Force needed = {F} N")
```

```python
import numpy as np

import matplotlib.pyplot as plt



ts = np.linspace(0, 5, 100)

plt.plot(ts, g * ts, label="v = gt")

plt.xlabel("time (s)")

plt.ylabel("velocity (m/s)")

plt.title("Velocity–time graph for free fall")

plt.grid(True)

plt.legend()

plt.show()
```
