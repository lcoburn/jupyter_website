# Graphs are representations, not pictures

This notebook uses one simple projectile idea to show how the **same motion** can look very different depending on:

1. the scale used on the axes,
2. the units used on the axes,
3. which variables we choose to plot.

The animations are designed for classroom discussion rather than calculation.

```python
import numpy as np

import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation

from IPython.display import HTML, display



g = 9.81



def projectile_between_people(x_end=20.0, y_peak=5.0, n_frames=100):

    """Return t, x, y for a projectile launched from y=0 and landing at x=x_end."""

    vy0 = np.sqrt(2 * g * y_peak)

    total_time = 2 * vy0 / g

    vx0 = x_end / total_time



    t = np.linspace(0, total_time, n_frames)

    x = vx0 * t

    y = vy0 * t - 0.5 * g * t**2

    y[y < 0] = 0

    return t, x, y



def show_animation(anim):

    plt.close(anim._fig)

    return HTML(anim.to_jshtml())
```

## 1. Same throw, different axis scales

The ball follows exactly the same physical trajectory in both panels.

- **Left:** equal spatial scale on both axes.
- **Right:** the axes are stretched differently.

Question for students: **Did the motion change, or only the way we drew it?**

```python
t, x, y = projectile_between_people(x_end=20, y_peak=5, n_frames=90)



fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4.5))



for ax in (ax1, ax2):

    ax.plot(x, y, alpha=0.35)

    ax.set_xlabel("horizontal position x (m)")

    ax.set_ylabel("height y (m)")

    ax.grid(alpha=0.2)



ax1.set_title("Equal spatial scale")

ax1.set_aspect("equal", adjustable="box")

ax1.set_xlim(-1, 21)

ax1.set_ylim(0, 6)



ax2.set_title("Same motion, unequal axis scales")

ax2.set_xlim(-1, 21)

ax2.set_ylim(0, 6)

# Deliberately do NOT force equal aspect ratio here.



dot1, = ax1.plot([], [], "o")

dot2, = ax2.plot([], [], "o")



def update_1(i):

    dot1.set_data([x[i]], [y[i]])

    dot2.set_data([x[i]], [y[i]])

    return dot1, dot2



anim1 = FuncAnimation(fig, update_1, frames=len(t), interval=45, blit=True)

show_animation(anim1)
```

## 2. Same kind of throw, but mixed units

Now the two people are only **2 m apart**, while the ball rises much higher.

In the second panel:

- horizontal position is shown in **centimetres**,
- vertical position is shown in **metres**.

The graph can look completely different even though the underlying motion has not changed.

Question for students: **What does the shape of the graph actually tell us if the axes use different units?**

```python
t2, x2_m, y2 = projectile_between_people(x_end=2, y_peak=20, n_frames=100)

x2_cm = 100 * x2_m



fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))



ax1.plot(x2_m, y2, alpha=0.35)

ax1.set_title("Physical position: metres on both axes")

ax1.set_xlabel("x (m)")

ax1.set_ylabel("y (m)")

ax1.set_xlim(-0.2, 2.2)

ax1.set_ylim(0, 21)

ax1.set_aspect("equal", adjustable="box")

ax1.grid(alpha=0.2)



ax2.plot(x2_cm, y2, alpha=0.35)

ax2.set_title("Same throw: x in cm, y in m")

ax2.set_xlabel("x (cm)")

ax2.set_ylabel("y (m)")

ax2.set_xlim(-10, 210)

ax2.set_ylim(0, 21)

ax2.grid(alpha=0.2)



dot1, = ax1.plot([], [], "o")

dot2, = ax2.plot([], [], "o")



def update_2(i):

    dot1.set_data([x2_m[i]], [y2[i]])

    dot2.set_data([x2_cm[i]], [y2[i]])

    return dot1, dot2



anim2 = FuncAnimation(fig, update_2, frames=len(t2), interval=45, blit=True)

show_animation(anim2)
```

## 3. Throw the ball straight up

If the ball goes straight up and comes straight back down:

- horizontal position barely changes,
- so plotting **y against x** wastes the horizontal axis,
- the upward and downward journeys lie on top of one another.

Instead, use **time** on the horizontal axis.

Now the graph separates the whole journey.

Question for students: **Which graph contains more useful information about this motion?**

```python
y_peak = 20

vy0 = np.sqrt(2 * g * y_peak)

total_time = 2 * vy0 / g



t3 = np.linspace(0, total_time, 100)

x3 = np.zeros_like(t3)

y3 = vy0 * t3 - 0.5 * g * t3**2

y3[y3 < 0] = 0



fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))



ax1.plot(x3, y3, alpha=0.35)

ax1.set_title("Position in space: y against x")

ax1.set_xlabel("x (m)")

ax1.set_ylabel("y (m)")

ax1.set_xlim(-1, 1)

ax1.set_ylim(0, 21)

ax1.grid(alpha=0.2)



ax2.plot(t3, y3, alpha=0.35)

ax2.set_title("Same motion: y against time")

ax2.set_xlabel("time t (s)")

ax2.set_ylabel("height y (m)")

ax2.set_xlim(0, total_time)

ax2.set_ylim(0, 21)

ax2.grid(alpha=0.2)



dot1, = ax1.plot([], [], "o")

dot2, = ax2.plot([], [], "o")



def update_3(i):

    dot1.set_data([x3[i]], [y3[i]])

    dot2.set_data([t3[i]], [y3[i]])

    return dot1, dot2



anim3 = FuncAnimation(fig, update_3, frames=len(t3), interval=45, blit=True)

show_animation(anim3)
```

## Teaching point

A graph is not necessarily a picture of what happened.

A graph is a **relationship between the variables chosen for its axes**.

Before interpreting any graph:

1. read the horizontal axis,
2. read the vertical axis,
3. check the units and scale,
4. only then interpret the shape.
