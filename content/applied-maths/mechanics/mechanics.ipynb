```python
# Constant acceleration: s = u*t + 0.5*a*t^2

u = 5.0   # initial velocity (m/s)

a = 2.0   # acceleration (m/s^2)

t = 4.0   # time (s)



s = u * t + 0.5 * a * t**2

print(f"After {t} s the displacement is {s} m")
```

```python
# Final velocity: v = u + a*t

v = u + a * t

print(f"Final velocity = {v} m/s")
```

```python
import numpy as np

import matplotlib.pyplot as plt



ts = np.linspace(0, 6, 100)

ss = u * ts + 0.5 * a * ts**2



plt.plot(ts, ss)

plt.xlabel("time (s)")

plt.ylabel("displacement (m)")

plt.title("Displacement–time graph (constant acceleration)")

plt.grid(True)

plt.show()
```
