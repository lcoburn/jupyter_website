```python
# Converting km/h to m/s: divide by 3.6
speed_kmh = 90
speed_ms = speed_kmh / 3.6
print(f"{speed_kmh} km/h = {speed_ms} m/s")
```

```python
# A journey: distance from speed and time (watch the units!)
time_minutes = 15
time_s = time_minutes * 60
distance_m = speed_ms * time_s
print(f"In {time_minutes} minutes at {speed_ms} m/s you travel {distance_m} m = {distance_m/1000} km")
```

```python
import matplotlib.pyplot as plt

speeds_kmh = [10, 30, 50, 70, 90, 120]
speeds_ms = [s / 3.6 for s in speeds_kmh]

plt.plot(speeds_kmh, speeds_ms, marker="o")
plt.xlabel("Speed (km/h)")
plt.ylabel("Speed (m/s)")
plt.title("Converting km/h to m/s")
plt.grid(True)
plt.show()
```
